Create two tensors:

 <br>

Compute gradients using backward(). <br>
Print gradients of w and b. <br>

Write full code.

## Write Answer 05:
We define parameters w and b, compute a linear model, apply sigmoid, then manually compute Binary Cross Entropy (BCE), followed by backpropagation.

```python
import torch

w = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)

x = torch.tensor(3.0)
y_true = torch.tensor(1.0)

z = w * x + b
y_pred = torch.sigmoid(z)

epsilon = 1e-7
loss = -(y_true * torch.log(y_pred + epsilon) + (1 - y_true) * torch.log(1 - y_pred + epsilon))

loss.backward()

print(w.grad)
print(b.grad)
```

# Question 07: [ Marks 20 ]


Create tensor w = 2.0 (requires_grad=True). <br>

Compute two different losses sequentially and call backward() twice. <br>
Print w.grad. <br>

Do not clear gradients. <br>
Explain what you observe. <br>

Write full code.


## Write Answer 07:



```python
import torch

w = torch.tensor(2.0, requires_grad=True)
loss1 = (w - 3) ** 2
loss1.backward()

loss2 = (w + 1) ** 2
loss2.backward()

print(w.grad)
```

When backward() is called multiple times without clearing gradients, PyTorch accumulates gradients in w.grad. This is why the final gradient is the sum of both individual gradients.
